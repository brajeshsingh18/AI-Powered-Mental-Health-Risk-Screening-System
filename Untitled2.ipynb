```python
feature_map = {
    "Gender": "gender",
    "Age": "age",
    "City": "living environment",
    "Profession": "profession",
    "Academic Pressure": "high academic pressure",
    "Work Pressure": "high work pressure",
    "CGPA": "low academic performance",
    "Study Satisfaction": "low study satisfaction",
    "Job Satisfaction": "low job satisfaction",
    "Sleep Duration": "poor sleep",
    "Dietary Habits": "poor diet",
    "Degree": "educational background",
    "Have you ever had suicidal thoughts ?": "having suicidal thoughts",
    "Work/Study Hours": "long working or study hours",
    "Financial Stress": "high financial stress",
    "Family History of Mental Illness": "family history of mental illness"
}

def get_original_feature(feature):
    if "_" in feature:
        return feature.rsplit("_", 1)[0]
    return feature


def generate_explanation(important_features, feature_map):
    explanations = []
    seen = set()

    for feature, value in important_features:
        if value > 0:
            original_feature = feature.rsplit("_", 1)[0]
            clean_name = feature_map.get(original_feature, original_feature.lower())

            # avoid duplicates
            if clean_name not in seen:
                explanations.append(clean_name)
                seen.add(clean_name)

    if not explanations:
        explanations.append("no strong risk factors")

    return explanations

```

```python
x_full = pd.concat([x_train, x_test]).sort_index()
```

```python
import shap

explainer = shap.Explainer(clf, x_train)
input_df = x_full

shap_values = explainer(input_df)

values_all = shap_values.values
features = x_train.columns
```

```python
all_explanations = []

for i in range(len(x_full)):
    values = values_all[i]

    important_features = []
    seen_features = set()

    top_indices = np.argsort(np.abs(values))[::-1] 

    for j in top_indices:
        feature_name = features[j]
        original_feature = feature_name.rsplit("_", 1)[0]

        if original_feature not in seen_features:
            important_features.append((feature_name, values[j]))
            seen_features.add(original_feature)

        if len(important_features) == 3:
            break
    important_features = sorted(important_features, key=lambda x: abs(x[1]), reverse=True)

    explanation = generate_explanation(important_features, feature_map)

    explanation_text = ", ".join(explanation)

    all_explanations.append(explanation_text)
```
